# Phase 2.2: PubMed PMID Acquisition

Run the rendered PubMed searches and save all PMIDs. This step is resumable: completed query IDs are skipped.


```python
from pathlib import Path
import sys

from IPython.display import display

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "pathogen_atlas_v2").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from pathogen_atlas_v2 import notebook_utils as nbutils
```

```python
NOTEBOOK_CONFIG_PATH = "configs/notebook_parameters.yaml"
PARAMS = nbutils.notebook_phase_parameters("phase2_pmid_acquisition", NOTEBOOK_CONFIG_PATH)

TEMPLATE_CONFIG_PATH = PARAMS["template_config_path"]
FILTER_CONFIG_PATH = PARAMS["filter_config_path"]
FORCE_REFRESH = bool(PARAMS["force_refresh"])
RUN_PUBMED_SEARCH = bool(PARAMS["run_pubmed_search"])
LOG_EVERY = int(PARAMS["log_every"])
```

```python
phase2_paths = nbutils.phase2_output_paths(TEMPLATE_CONFIG_PATH)
display(nbutils.existing_outputs(phase2_paths))
display(nbutils.phase2_pmid_acquisition_preflight(TEMPLATE_CONFIG_PATH, FILTER_CONFIG_PATH, force_refresh=FORCE_REFRESH))
```

```python
if RUN_PUBMED_SEARCH:
    pmid_result = nbutils.run_phase_2_pmid_acquisition(
        TEMPLATE_CONFIG_PATH,
        FILTER_CONFIG_PATH,
        force_refresh=FORCE_REFRESH,
        log_every=LOG_EVERY,
    )
else:
    pmid_result = {"status": "not_run", "set_RUN_PUBMED_SEARCH": True}

pmid_summary = nbutils.phase_run_summary(pmid_result)
display(pmid_summary["overview"])
display(pmid_summary["counts"])
```
